```python
import pandas as pd
import glob

# Get all filtered parquet files
files = sorted(glob.glob('Yield_*_filtered.parquet'))
print(f"Found {len(files)} files: {files}\n")

# Read all files and concatenate
dfs = []
for file in files:
    df = pd.read_parquet(file)
    dfs.append(df)

all_data = pd.concat(dfs, ignore_index=True)
all_data = all_data[all_data['Yield'] > 29]
print(f"{'='*60}")
print(f"Overall Statistics (All Years Combined)")
print(f"{'='*60}")
print(f"Total number of records: {len(all_data)}")

# Get numeric columns
numeric_cols = all_data.select_dtypes(include=['number']).columns

if len(numeric_cols) > 0:
    overall_stats = all_data[numeric_cols].agg(['mean', 'min', 'max', 'std'])
    print(f"\nOverall numeric columns statistics:")
    print(overall_stats)
else:
    print("No numeric columns found")

# Statistics by crop
if 'Crop' in all_data.columns:
    # print(f"\n\n{'='*60}")
    # print(f"Statistics by Crop Type")
    # print(f"{'='*60}")
    
    crops = all_data['Crop'].unique()
    print(f"Found {len(crops)} unique crops: {crops}\n")
    
    for crop in sorted(crops):
        crop_df = all_data[all_data['Crop'] == crop]
        print(f"\n{'-'*60}")
        print(f"Crop: {crop}")
        print(f"{'-'*60}")
        print(f"Number of records: {len(crop_df)}")
        
        if len(numeric_cols) > 0:
            crop_stats = crop_df[numeric_cols].agg(['mean', 'min', 'max', 'std'])
            print(f"\nNumeric columns statistics:")
            print(crop_stats)
else:
    print("\n\nNo 'Crop' column found in the data")
```
